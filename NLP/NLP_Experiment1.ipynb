```python
import nltk
nltk.download('punkt_tab')
```

```python
sample_text = "I don't want other people to mention my self. Besides if I can be changed with a few words, then that wouldn't be my 'self' would it?"
```

```python
sentence_tokanized = nltk.sent_tokenize(sample_text)
sentence_tokanized
```

```python
all_word_tokens = []
for sentence in sentence_tokanized:
  word_tokenized = nltk.word_tokenize(sentence)
  all_word_tokens.append(word_tokenized)
all_word_tokens
```

```python
import nltk

def tokenizer(para):
  sentence_tokanize = nltk.sent_tokenize(para)
  print(f"The sentences = {sentence_tokanize}" )
  print(f"The Count of Sentences = {len(sentence_tokanize)}")
  word_tokenized = nltk.word_tokenize(para)
  print(f"The words = {word_tokenized}" )
  print(f"The Count of Words = {len(word_tokenized)}")
```

```python
tokenizer("As a veteran on this battlefield of life, I've gotten used to losing. That's why I always hate nice girls")
```
