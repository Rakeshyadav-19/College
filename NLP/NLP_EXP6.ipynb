Title: Implement sentiment analysis using Natural Language Processing (NLP) techniques.

```python
# Import required libraries
import pandas as pd
import string
import nltk
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Download stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
```

```python
# Step 1: Sample Dataset
data = {
    "text": [
        "The movie was an absolute masterpiece, I was captivated from start to finish.", # positive
        "This book is a total waste of time, the plot is nonsensical and the characters are flat.", # negative
        "The restaurant had decent food, but the service was incredibly slow.", # neutral
        "I had a fantastic experience at the concert, the band was amazing!", # positive
        "The new software update is full of bugs and crashes constantly.", # negative
        "The painting was interesting, though not particularly memorable.", # neutral
        "What a breathtaking view from the mountain top!", # positive
        "I regret buying this product, it broke after just one use.", # negative
        "The lecture was informative and well-presented.", # positive
        "The traffic was terrible this morning, I was late for work.", # negative
    ],
    "label": [
        "positive", "negative", "neutral", "positive", "negative", "neutral", "positive", "negative", "positive", "negative"
    ]
}

df = pd.DataFrame(data)
```

```python
# Step 2: Text Preprocessing
def preprocess(text):
    text = text.lower()                                    # Lowercasing
    text = "".join([c for c in text if c not in string.punctuation])  # Remove punctuation
    words = [w for w in text.split() if w not in stop_words]          # Remove stopwords
    return " ".join(words)

df["clean_text"] = df["text"].apply(preprocess)

print("Preprocessed Data:\n", df)
```

```python
# Step 3: Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    df["clean_text"], df["label"], test_size=0.3, random_state=42
)
```

```python
# Step 4: Bag-of-Words (BoW)
# ------------------------------
count_vect = CountVectorizer()
X_train_bow = count_vect.fit_transform(X_train)
X_test_bow = count_vect.transform(X_test)

# Train Naive Bayes model
model_bow = MultinomialNB()
model_bow.fit(X_train_bow, y_train)
y_pred_bow = model_bow.predict(X_test_bow)

print("\nBag-of-Words Results:")
print("Accuracy:", accuracy_score(y_test, y_pred_bow))
print(classification_report(y_test, y_pred_bow))
```

```python
# Step 5: TF-IDF
# ------------------------------
tfidf_vect = TfidfVectorizer()
X_train_tfidf = tfidf_vect.fit_transform(X_train)
X_test_tfidf = tfidf_vect.transform(X_test)

# Train Naive Bayes model
model_tfidf = MultinomialNB()
model_tfidf.fit(X_train_tfidf, y_train)
y_pred_tfidf = model_tfidf.predict(X_test_tfidf)

print("\nTF-IDF Results:")
print("Accuracy:", accuracy_score(y_test, y_pred_tfidf))
print(classification_report(y_test, y_pred_tfidf))
```
