```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
# Download necessary data
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# Sample text
text = "A balanced lifestyle is crucial for overall health and well-being. It encompasses a combination of nutritious eating habits, regular physical activity, adequate sleep, and effective stress management."
```

Activity 2:Apply tokenization, stop word removal, and stemming.

```python
# Sample text
text = "A balanced lifestyle is crucial for overall health and well-being. It encompasses a combination of nutritious eating habits, regular physical activity, adequate sleep, and effective stress management."
# Tokenization
sentences = sent_tokenize(text)
words = word_tokenize(text)
print("Sentences:", sentences)
print("Words:", words)
```

```python
# Stop word removal
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.lower() not in stop_words]
print("Filtered Words:", filtered_words)
```

```python
# Stemming
ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]
print ("Stemmed Words:", stemmed_words)
```

Activity 3: Count the number of stop words and stemmed words.

```python
# Count stop words
stopword_count = sum(1 for word in sentences if word in stop_words)

# Apply stemming
stemmed_words = [ps.stem(word) for word in sentences]
stemmed_count = len(set(stemmed_words))  # unique stemmed words

print("Total words:", len(sentences))
print("Stop words count:", stopword_count)
print("Unique stemmed words count:", stemmed_count)
```
