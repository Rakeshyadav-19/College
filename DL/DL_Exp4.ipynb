```python
import speech_recognition as sr
from transformers import pipeline
```

```python
# Initialize speech recognizer and NLP model
recognizer = sr.Recognizer()
nlp = pipeline("text-classification", model="distilbert-base-uncased-finetuned-sst-2-english")


# Function to capture voice input and convert it to text
def listen_to_command():
    with sr.Microphone() as source:
        print("Listening for a command...")
        recognizer.adjust_for_ambient_noise(source)
        audio = recognizer.listen(source)
    try:
        command = recognizer.recognize_google(audio)
        print(f"Recognized Command: {command}")
        return command.lower()
    except sr.UnknownValueError:
        print("Sorry, I did not understand that.")
    except sr.RequestError:
        print("Sorry, there was a problem with the request.")
    return None
```

```python
# Function to process command using NLP and execute actions
def process_command(command):
    if command:
        # Simple NLP-based intent detection
        if "light" in command:
            if "on" in command:
                print("Turning on the lights.")
                # Code to turn on lights
            elif "off" in command:
                print("Turning off the lights.")
                # Code to turn off lights
        elif "fan" in command:
            if "on" in command:
                print("Turning on the fan.")
                # Code to turn on fan
            elif "off" in command:
                print("Turning off the fan.")
                # Code to turn off fan
        elif "temperature" in command:
            print("Adjusting the temperature.")
            # Code to adjust temperature
        else:
            print("Sorry, I couldn't understand the command.")
    else:
        print("No command detected.")
# Main loop to continuously listen and respond to commands
while True:
    command = listen_to_command()
    process_command(command)
```
