```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel, set_seed
import torch

model_name = "gpt2"

tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu") # Fix: Set device to "cpu" if cuda is not available
model.to(device)

prompt = "Wake up to reality!"
inputs = tokenizer(prompt, return_tensors="pt").to(device) # Add input tensor and move it to the selected device

output_ids = model.generate(  **inputs,  max_new_tokens=30 )

generated_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)
print("=== Generated Text ===")
print(generated_text)
```
