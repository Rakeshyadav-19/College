```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
```

```python
# Load an image
image = cv2.imread('/content/ken.jpg')
```

```python
# Rotate Image
(h, w) = image.shape[:2]
center = (w // 2, h // 2)
matrix = cv2.getRotationMatrix2D(center, 45, 1.0)
rotated_img = cv2.warpAffine(image, matrix, (w, h))
```

```python
# Translate Image
matrix = np.float32([[1, 0, 50], [0, 1, 30]])
translated_img = cv2.warpAffine(image, matrix, (w, h))
```

```python
# Scale Image
scaled_img = cv2.resize(image, None, fx=1.5, fy=1.5, interpolation=cv2.INTER_LINEAR)
```

```python
import matplotlib.pyplot as plt
# Convert images from BGR to RGB
rotated_img_rgb = cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB)
translated_img_rgb = cv2.cvtColor(translated_img, cv2.COLOR_BGR2RGB)
scaled_img_rgb = cv2.cvtColor(scaled_img, cv2.COLOR_BGR2RGB)

# Display images using Matplotlib
plt.figure(figsize=(10, 5))  # Adjust figure size as needed

plt.subplot(1, 3, 1)
plt.imshow(rotated_img_rgb)
plt.title('Rotated Image')

plt.subplot(1, 3, 2)
plt.imshow(translated_img_rgb)
plt.title('Translated Image')

plt.subplot(1, 3, 3)
plt.imshow(scaled_img_rgb)
plt.title('Scaled Image')

plt.tight_layout()  # Adjust layout for better spacing
plt.show()
```
