```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
```

```python
# Load the image
image = cv2.imread('/content/Manga-Shrine.png', cv2.IMREAD_GRAYSCALE)
# Check if image is loaded
if image is None:
  print("Error: Could not load image.")
  return
```

```python
# Apply edge detection methods
# Define lower and upper thresholds for Canny edge detection
lower_threshold = 50  # Example value, adjust as needed
upper_threshold = 150 # Example value, adjust as needed

edges_canny = cv2.Canny(image, lower_threshold, upper_threshold)
edges_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=3)
edges_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=3)
edges_sobel = cv2.magnitude(edges_x, edges_y)
edges_laplacian = cv2.Laplacian(image, cv2.CV_64F)
```

```python
# Apply edge detection methods
lower_threshold = 50  # Adjust as needed
upper_threshold = 150 # Adjust as needed

edges_canny = cv2.Canny(image, lower_threshold, upper_threshold)
edges_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=3)
edges_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=3)
edges_sobel = cv2.magnitude(edges_x, edges_y)
edges_sobel = np.uint8(np.absolute(edges_sobel)) # Convert to uint8 for display
edges_laplacian = cv2.Laplacian(image, cv2.CV_64F)
edges_laplacian = np.uint8(np.absolute(edges_laplacian)) # Convert to uint8 for display
```

```python
# Display the original and edge-detected images using matplotlib
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(edges_canny, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(edges_sobel, cmap='gray')
plt.title('Sobel Edge Detection')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(edges_laplacian, cmap='gray')
plt.title('Laplacian Edge Detection')
plt.axis('off')

plt.show()
```
