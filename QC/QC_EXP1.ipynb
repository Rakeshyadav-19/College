```python
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_vector
import matplotlib.pyplot as plt
import numpy as np
```

```python
def simulate_gate(gate_name):
  qc=QuantumCircuit(1)

  if gate_name=='X':
    qc.x(0)
  elif gate_name=='Y':
    qc.y(0)
  elif gate_name=='Z':
    qc.z(0)
  elif gate_name=='H':
    qc.h(0)

  sv = Statevector.from_instruction(qc)
  print(f"{gate_name}, Gate Applied:\n {sv} ")

  bloch = sv.data
  plot_bloch_vector([np.real(bloch[0]*np.conj(bloch[1]) + bloch[1]*np.conj(bloch[0])),np.imag(bloch[0]*np.conj(bloch[1]) - bloch[1]*np.conj(bloch[0])),np.abs(bloch[0])*2 - np.abs(bloch[1])*2])
  plt.title(f"Bloch Sphere after {gate_name} gate")
  plt.show()

for gate in ['X', 'Y', 'Z', 'H']:
  simulate_gate(gate)
```
