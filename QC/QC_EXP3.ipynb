```python
def constant_oracle(qc):
    # Does nothing — output remains 0
    pass
def balanced_oracle(qc):
    qc.cx(0, 1)  # Controlled NOT
```

```python
# Create 2-qubit, 1-classical-bit circuit
qc = QuantumCircuit(2, 1)

# Initialize second qubit to |1⟩
qc.x(1)

# Apply Hadamard to both qubits
qc.h(0)
qc.h(1)

# Apply oracle
balanced_oracle(qc)   # or constant_oracle(qc)

# Apply Hadamard to first qubit
qc.h(0)

# Measure first qubit
qc.measure(0, 0)

from qiskit.visualization.circuit.matplotlib import MatplotlibDrawer
qc.draw('text')
```
